```python
import numpy as np
import matplotlib.pyplot as plt
```

Cell 2 – Define the function and its derivative

```python
# Define the function
def function(x):
    return (x + 3)**2

# Define the derivative
def derivative(x):
    return 2 * (x + 3)
```

Cell 3 – Initialize parameters

```python
# Starting point
x = 2

# Learning rate
learning_rate = 0.1

# Number of iterations
iterations = 50

# Store values
x_values = []
y_values = []
```

Cell 4 – Implement Gradient Descent Algorithm

```python
for i in range(iterations):

    # Calculate function value
    y = function(x)

    # Store values
    x_values.append(x)
    y_values.append(y)

    # Calculate gradient
    grad = derivative(x)

    # Update x
    x = x - learning_rate * grad

    print("Iteration:", i + 1, "x =", x, "y =", function(x))

print("\nLocal Minimum Point:", x)
print("Minimum Value:", function(x))
```

Cell 5 – Plot the function

```python
# Generate x values
x_plot = np.linspace(-8, 5, 100)

# Calculate y values
y_plot = function(x_plot)

# Plot function
plt.figure(figsize=(8, 5))

plt.plot(x_plot, y_plot, label="y = (x+3)^2")

# Plot gradient descent points
plt.scatter(x_values, y_values, color="red",
            label="Gradient Descent")

plt.title("Gradient Descent Algorithm")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid()

plt.show()
```

Cell 6 – Display final result

```python
print("Starting Point: x = 2")
print("Local Minimum x =", round(x, 6))
print("Minimum y =", round(function(x), 6))
```
